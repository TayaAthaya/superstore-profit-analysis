Analisis difokuskan pada penjualan, profit, profit margin, pola diskon, dan data

## Profit by Product Category

This analysis aims to identify which product categories contribute the most
to overall business profitability. By grouping data by category and summing
the profit, we can evaluate business performance beyond sales volume.

Analisis ini bertujuan untuk mengetahui kategori produk mana yang memberikan
kontribusi keuntungan terbesar bagi bisnis. Dengan mengelompokkan data
berdasarkan kategori dan menjumlahkan profit, kita dapat menilai performa
bisnis secara lebih akurat, bukan hanya dari sisi penjualan.


```python
profit_by_category = df.groupby('Category')['Profit'].sum()
profit_by_category
```



---



## Sales by Product Category

This section analyzes total sales for each product category.
The result is later compared with profit data to identify categories
with high sales but relatively low profitability.

Pada bagian ini, dilakukan analisis total penjualan untuk setiap kategori produk.
Hasil ini kemudian dibandingkan dengan data profit untuk mengidentifikasi
kategori yang memiliki penjualan tinggi tetapi keuntungan yang tidak sebanding.


```python
sales_by_category = df.groupby('Category')['Sales'].sum()
sales_by_category
```



---



## Category Performance Analysis: Sales, Profit, and Profit Margin

In this section, we combine total sales and total profit for each product category
into a single table. We then calculate the profit margin to better understand
how efficiently each category generates profit relative to its sales.

This analysis helps identify categories that may appear strong based on sales
performance but are actually less profitable when profitability efficiency
is taken into account.

Pada bagian ini, data total penjualan (sales) dan total keuntungan (profit)
untuk setiap kategori digabungkan ke dalam satu tabel. Selanjutnya, dilakukan
perhitungan profit margin untuk melihat seberapa efisien setiap kategori dalam
menghasilkan keuntungan dibandingkan dengan nilai penjualannya.

Analisis ini bertujuan untuk mengidentifikasi kategori yang terlihat kuat dari
sisi penjualan, namun sebenarnya kurang menguntungkan dari sisi profitabilitas.


```python
category_performance = pd.DataFrame({
    'Sales': sales_by_category,
    'Profit': profit_by_category
})

category_performance['Profit Margin (%)'] = (category_performance['Profit'] / category_performance['Sales']) * 100

# Sort by sales in descending order and profit margin in ascending order to easily spot high sales, low profit
category_performance_sorted = category_performance.sort_values(by=['Sales', 'Profit Margin (%)'], ascending=[False, True])

display(category_performance_sorted)
```



---



## Average Discount by Category

In this section, we analyze the average discount applied to each product category.
The goal is to verify whether categories with lower profit margins are associated
with higher discount levels.

This analysis helps validate whether discounting strategy is a key factor
contributing to low profitability.

Pada bagian ini, dilakukan analisis rata-rata diskon untuk setiap kategori produk.
Tujuannya adalah untuk membuktikan apakah kategori dengan profit margin rendah
memang memiliki tingkat diskon yang lebih tinggi.

Analisis ini digunakan untuk menguji apakah strategi pemberian diskon menjadi
faktor utama penyebab rendahnya profitabilitas.


```python
avg_discount_by_category = df.groupby('Category')['Discount'].mean()
avg_discount_by_category
```



---



## Furniture Sub-Category Profit Analysis

In this section, we drill down into the Furniture category to analyze profit
performance at the sub-category level. The objective is to identify which
Furniture sub-categories contribute the most to losses or low profitability.

This deeper analysis helps pinpoint specific problem areas within the Furniture
category that may require changes in pricing, discount strategy, or inventory
management.

Pada bagian ini, analisis difokuskan pada kategori Furniture hingga ke level
sub-category untuk melihat performa profit secara lebih detail. Tujuannya adalah
mengidentifikasi sub-category Furniture yang paling berkontribusi terhadap
rendahnya profit atau bahkan mengalami kerugian.

Analisis ini membantu menentukan area spesifik yang bermasalah dan dapat menjadi
dasar pengambilan keputusan bisnis, seperti penyesuaian diskon, harga, atau jumlah
stok.


```python
# Filter data to Furniture category only
furniture_df = df[df['Category'] == 'Furniture']

# Calculate total profit per Furniture sub-category
profit_by_furniture_subcat = (
    furniture_df
    .groupby('Sub-Category')['Profit']
    .sum()
    .sort_values()
)

profit_by_furniture_subcat
```



---

Analisis ini dilakukan pada level transaksi untuk memastikan bahwa strategi

```python
# Filter data to Tables sub-category only
tables_df = df[
    (df['Category'] == 'Furniture') &
    (df['Sub-Category'] == 'Tables')
]

# Descriptive statistics for Discount and Profit
tables_df[['Discount', 'Profit']].describe()
```

## Final Conclusion

This analysis reveals that high sales performance does not necessarily translate
into high profitability. Among all product categories, Furniture stands out as
a category with relatively high sales but significantly low profit margin.

A deeper analysis shows that the primary source of this issue comes from the
Tables sub-category. Tables consistently generate negative profit, driven by
aggressive discounting strategies. Transaction-level analysis confirms that
most Tables sales involve high discounts (averaging around 26%), resulting in
systematically negative profit despite steady demand.

These findings indicate that the losses in Furniture are structural rather than
incidental, and are largely caused by pricing and discount policies rather than
lack of sales volume.

Analisis ini menunjukkan bahwa tingginya penjualan tidak selalu berarti tingginya
keuntungan. Di antara seluruh kategori produk, Furniture terlihat memiliki nilai
penjualan yang cukup besar, namun profit margin yang sangat rendah.

Analisis lebih lanjut mengungkap bahwa sumber utama permasalahan berasal dari
sub-category Tables. Tables secara konsisten menghasilkan profit negatif akibat
strategi diskon yang terlalu agresif. Analisis pada level transaksi membuktikan
bahwa sebagian besar penjualan Tables melibatkan diskon tinggi (rata-rata sekitar
26%), sehingga menyebabkan kerugian yang bersifat sistemik meskipun permintaan
pasar cukup stabil.

Temuan ini menunjukkan bahwa kerugian pada kategori Furniture bukan disebabkan oleh
rendahnya penjualan, melainkan oleh kebijakan harga dan diskon yang kurang optimal.


## Business Recommendations

Based on the analysis, it is recommended to optimize discount strategies for the
Furniture category, particularly for the Tables sub-category. Rather than relying
on aggressive discounting to drive sales, the business should evaluate more
selective or gradual discount approaches to protect profit margins.

Possible actions include testing lower discount thresholds for Tables, limiting
high discounts to specific periods or customer segments, and closely monitoring
profit impact after discount adjustments. These measures can help improve overall
profitability without significantly reducing sales volume.

Berdasarkan hasil analisis, disarankan untuk melakukan optimasi strategi diskon
pada kategori Furniture, khususnya sub-category Tables. Alih-alih mengandalkan
diskon besar untuk mendorong penjualan, bisnis perlu menerapkan pendekatan diskon
yang lebih selektif atau bertahap guna menjaga margin keuntungan.

Beberapa langkah yang dapat dipertimbangkan antara lain menguji batas diskon yang
lebih rendah untuk produk Tables, membatasi diskon besar pada periode atau segmen
pelanggan tertentu, serta memantau dampak perubahan diskon terhadap profit secara
berkelanjutan. Pendekatan ini diharapkan dapat meningkatkan profitabilitas tanpa
mengorbankan volume penjualan secara signifikan.
